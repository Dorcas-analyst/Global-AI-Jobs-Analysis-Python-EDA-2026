# CREATING SEGMENTS AND VISUALIZATIONS

```python
segments = {
    'Compensation & Finance': ['salary_usd', 'bonus_usd', 'salary_percentile',
                               'tax_rate_percent', 'cost_of_living_index'],
    'Wellbeing & Work-Life': ['work_life_balance_score', 'employee_satisfaction',
                              'job_security_score', 'layoff_risk',
                              'weekly_hours', 'vacation_days'],
    'Career Growth': ['career_growth_score', 'promotion_speed', 'experience_years',
                      'skill_demand_score', 'automation_risk'],
    'Hiring & Market': ['interview_rounds', 'hiring_difficulty_score', 'job_openings',
                        'offer_acceptance_rate', 'company_rating'],
    'Company & Economy': ['company_funding_billion', 'ai_adoption_score',
                          'ai_maturity_years', 'economic_index'],
}

def plot_segment(df, title, cols, ncols=3):
    nrows = -(-len(cols) // ncols)              # ceiling division
    fig, axes = plt.subplots(nrows, ncols, figsize=(5 * ncols, 4 * nrows))
    axes = axes.flatten()

    for ax, col in zip(axes, cols):
        if df[col].dtype == 'object':           # categorical: top 10 only
            top = df[col].value_counts().head(10)
            sns.barplot(x=top.values, y=top.index, ax=ax, color='navy')
        elif df[col].nunique() <= 15:           # discrete numeric: count each value
            sns.countplot(x=df[col], ax=ax, color='navy')
        else:                                   # continuous: histogram + curve
            sns.histplot(df[col], bins=20, kde=True, ax=ax, color='navy')
        ax.set_title(col)

    for ax in axes[len(cols):]:                 # hide unused slots
        ax.axis('off')

    fig.suptitle(title, fontsize=16)
    plt.tight_layout()
    plt.show()

for name, cols in segments.items():
    plot_segment(df, name, cols)
    plt.savefig("segments charts")
```

```python
work_mode_proportion = df['work_mode'].value_counts()
plt.figure(figsize=(6,6))

plt.pie(
    work_mode_proportion,
    labels = work_mode_proportion.index,
    autopct = '%1.1f%%',
    startangle = 90
)
plt.title('Proportion Of Work Mode')
plt.tight_layout()
plt.savefig("Porportion OF Work Mode")
plt.show()
```

```python
yearly = df.groupby('year')[['salary_usd', 'bonus_usd']].mean().sort_index()

fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.plot(yearly.index, yearly['salary_usd'], marker='o', color='navy', label='Avg Salary (USD)')
ax1.set_xlabel('Year')
ax1.set_ylabel('Average Salary (USD)', color='navy')
ax1.tick_params(axis='y', labelcolor='navy')

ax2 = ax1.twinx()
ax2.plot(yearly.index, yearly['bonus_usd'], marker='o', color='orange', label='Avg Bonus (USD)')
ax2.set_ylabel('Average Bonus (USD)', color='orange')
ax2.tick_params(axis='y', labelcolor='orange')

fig.suptitle('Average Salary and Bonus by Year')
fig.legend(loc='upper left', bbox_to_anchor=(0.1, 0.9))
plt.tight_layout()
plt.savefig("Average Salary and Bonus by Year chart")
plt.show()
```
